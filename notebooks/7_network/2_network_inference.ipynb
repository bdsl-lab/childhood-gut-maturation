# Network inference

## Control all -> MMI high -> MMI low -> PIBD

```r
high_net_genus <- netConstruct(high_genus,       # phyloseq
                          measure = "sparcc",    # Network inference measure: sparCC
                          sparsMethod = "threshold", 
                          thresh = 0.3,
                          set.seed(123456),
                          verbose = 3)

high_net_genus_pos <- high_net_genus
# 1. negative or 0 association add at remove_edge function
remove_edge <- high_net_genus_pos$assoMat1 <= 0
diag(remove_edge) <- FALSE
# 2. Association matrix
high_net_genus_pos$assoMat1[remove_edge] <- 0
# 3. Adjacency matrix
high_net_genus_pos$adjaMat1[remove_edge] <- 0
# 4. Similarity matrix
# Deleted edge is similarity = 0
high_net_genus_pos$simMat1[remove_edge] <- 0
# 5. Dissimilarity matrix
high_net_genus_pos$dissMat1[remove_edge] <- Inf
# 6. Edge list
high_net_genus_pos$edgelist1 <- subset(high_net_genus_pos$edgelist1,asso > 0)
# 7. Positive association network analysis
high_props_genus_pos <- netAnalyze(high_net_genus_pos,clustMethod = "cluster_fast_greedy")

network_high <- list(net_genus = high_net_genus, props_genus = high_props_genus)
saveRDS(network_high, file="data/3-results/5_network_analysis/1_netcomi_inference/high/high_network_nopos.rds")

network_high_pos <- list(net_genus = high_net_genus_pos, props_genus = high_props_genus_pos)
saveRDS(network_high_pos, file="data/3-results/5_network_analysis/1_netcomi_inference/high/high_network_pos.rds")
```
