```python
import pandas as pd
import numpy as np
```

```python
h_nodes = pd.read_csv("data/3_results/5_network_analysis/5_2_network_export/high/high_nodes_topology.csv", sep=',')
h_edges = pd.read_csv("data/3_results/5_network_analysis/5_2_network_export/high/high_edges.csv", sep=',')
h_re_ab = pd.read_csv("data/3_results/5_network_analysis/5_2_network_export/high/high_rel_abundance.csv", sep=',')
l_nodes = pd.read_csv("data/3_results/5_network_analysis/5_2_network_export/low/low_nodes_topology.csv", sep=',')
l_edges = pd.read_csv("data/3_results/5_network_analysis/5_2_network_export/low/low_edges.csv", sep=',')
l_re_ab = pd.read_csv("data/3_results/5_network_analysis/5_2_network_export/low/low_rel_abundance.csv", sep=',')
p_nodes = pd.read_csv("data/3_results/5_network_analysis/5_2_network_export/pibd/pibd_nodes_topology.csv", sep=',')
p_edges = pd.read_csv("data/3_results/5_network_analysis/5_2_network_export/pibd/pibd_edges.csv", sep=',')
p_re_ab = pd.read_csv("data/3_results/5_network_analysis/5_2_network_export/pibd/pibd_rel_abundance.csv", sep=',')
```

```python
roman_map = {
    1.0: "I",
    2.0: "II",
    3.0: "III",
    4.0: "IV",
    5.0: "V",
    6.0: "VI",
    7.0: "VII",
    8.0: "VIII",
    9.0: "IX",
    10.0: "X",
    11.0: "XI"
}


h_nodes["cluster"] = h_nodes["cluster"].replace(roman_map)
l_nodes["cluster"] = l_nodes["cluster"].replace(roman_map)
p_nodes["cluster"] = p_nodes["cluster"].replace(roman_map)
```

```python
h_re_ab
```

```python
keys = ["Label"]

h_renamed = h_nodes.rename(columns={
    "Id":"Id_high",
    "cluster":"cluster_high",
    "degree":"degree_high",
    "betweenness":"betweenness_high",
    "closeness":"closeness_high",
    "eigenvector":"eigenvector_high",
    "prevalence":"prevalence_high",
    "abundance":"abundance_high",
})

h_re_ab = h_re_ab.rename(columns={
    "relative_abundance":"high_rel"
})

l_renamed = l_nodes.rename(columns={
    "Id":"Id_low",
    "cluster":"cluster_low",
    "degree":"degree_low",
    "betweenness":"betweenness_low",
    "closeness":"closeness_low",
    "eigenvector":"eigenvector_low",
    "prevalence":"prevalence_low",
    "abundance":"abundance_low",
})

l_re_ab = l_re_ab.rename(columns={
    "relative_abundance":"low_rel"
})

p_renamed = p_nodes.rename(columns={
    "Id":"Id_pibd",
    "cluster":"cluster_pibd",
    "degree":"degree_pibd",
    "betweenness":"betweenness_pibd",
    "closeness":"closeness_pibd",
    "eigenvector":"eigenvector_pibd",
    "prevalence":"prevalence_pibd",
    "abundance":"abundance_pibd",
})

p_re_ab = p_re_ab.rename(columns={
    "relative_abundance":"pibd_rel"
})
```

```python
merged = h_renamed.merge(l_renamed, on=keys, how="outer")
```

```python
merged_final = merged.merge(p_renamed, on=keys, how="outer")
```

```python
merged_final.sort_values('Label', ascending=True)

merged_final
```

```python
# merged
merged_final["phylum"] = (
    merged_final["phylum_x"]
    .combine_first(merged_final["phylum_y"])
    .combine_first(merged_final["phylum"])
)

# conflict check
conflict_mask = (
    (merged_final["phylum_x"].notna()) &
    (merged_final["phylum_y"].notna()) &
    (merged_final["phylum_x"] != merged_final["phylum_y"])
)

if conflict_mask.any():
    print("phylum_x vs phylum_y mismatch detected")
    print(merged_final.loc[conflict_mask, ["Id", "phylum_x", "phylum_y"]].head())

# column delete
merged_final = merged_final.drop(columns=["phylum_x", "phylum_y"], errors="ignore")
```

```python
merged_final = merged_final.sort_values(by = ['Id_high'], ascending=True).copy()
```

```python
merged_final
```

```python
## Export to one of file

h_edges['group'] = "high_edge"

l_edges['group'] = "low_edge"

p_edges['group'] = "pibd_edge"

con_edges = pd.concat([h_edges, l_edges, p_edges])

con_edges
```
