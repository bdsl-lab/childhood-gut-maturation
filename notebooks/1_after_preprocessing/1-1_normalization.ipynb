## Normalization
- Prevalence > 10% filtering
    1. Control, PIBD feature table
    2. log1p normalization


```python
import pandas as pd
import numpy as np
```

```python
row = "data/2_preprocessed/ft.tsv"
meta_path = "data/pre_metadata.tsv"
meta = pd.read_csv(meta_path, sep='\t').set_index("#SampleID")
meta = meta.dropna(subset=["Months"])


genera = pd.read_csv(row, sep='\t', index_col=0).T


all = genera.index.intersection(meta.index)
genera = genera.loc[all]
meta = meta.loc[all]

## Only control

meta_1 = meta[(meta["Host_disease"] == "Control")]
common = genera.index.intersection(meta_1.index)

```

```python
meta.count()
```

### All samples prevalence 10% filtering

```python
## Prevalence > 10% filtering

all_pre10 = genera.loc[:, (genera > 0).mean() >= 0.1]
all_pre10_T = all_pre10.T ## Network input table generate

## Control selected

meta_1 = meta[(meta["Host_disease"] == "Control")]
common = genera.index.intersection(meta_1.index)
all_pre10_con = all_pre10.loc[common]

## PIBD(CD, UC) data selected
pibd = meta[(meta["Host_disease"] != "Control")]
pi = all_pre10.index.intersection(pibd.index)
all_pre10_pibd = all_pre10.loc[pi]



all_pre10_T.to_csv("data/2_preprocessed/prv10/prv10_network.tsv", sep='\t') ## Network import file
all_pre10.to_csv("data/2_preprocessed/prv10/prv10_ft.tsv", sep='\t') ## All group (HC, CD, UC)
all_pre10_con.to_csv("data/2_preprocessed/prv10/prv10_ft_control.tsv", sep='\t') ## control subset to prevalence > 10% filtering table
all_pre10_pibd.to_csv("data/2_preprocessed/prv10/prv10_ft_pibd.tsv", sep='\t') ## pibd subset to prevalence > 10% filtering table
```

```python
all_pre10 = genera.loc[:, (genera > 0).mean() >= 0.1]
```

```python
all_pre10
```

### log 1p + pre 10

```python
all_pre10_log = np.log1p(all_pre10)
all_pre10_log_con = all_pre10_log.loc[common]
all_pre10_log_pibd = all_pre10_log.loc[pi]

all_pre10_log.to_csv("data/2_preprocessed/log1p/prv10_ft_log1p.tsv", sep='\t') ## All group(HC, CD, UC)
all_pre10_log_con.to_csv("data/2_preprocessed/log1p/prv10_ft_log1p_control.tsv", sep='\t') ## Control subset to prevalence > 10% filtered log1p normalization feature table
all_pre10_log_pibd.to_csv("data/2_preprocessed/log1p/prv10_ft_log1p_pibd.tsv", sep='\t') ## PIBD subset to prevalence > 10% filtered log1p normalization feature table
```
